<b>Example 1</b>

```python
basket=['apple', 'orange', 'apple', 'pear', 'orange', 'banana']
fruits=set(basket) # create a set without duplicates
print("Fruits in the basket are:")
print(fruits)

print("'orange' in fruits:", 'orange' in fruits) # fast membership testing
print("'crabgrass' in fruits:", 'crabgrass' in fruits)
```

<b>Example 2</b>

```python
#Demonstrate set operations on unique letters from two words
a=set('abracadabra')
b=set('alacazam')
print(a)

a - b  # letters in a but not in b
print("Difference:")
print(a - b)
a | b  # letters in a or b or both
print("Union:")
print(a | b)
a & b  # letters in both a and b
print("Intersection:")
print(a & b)
a ^ b  # letters in a or b but not both
print("Symmetric Difference:")
print(a ^ b)
```
